```python
#   EJERCICIO NUMERO 1
import matplotlib.pyplot as plt

estudiantes = ['Ana', 'Luis', 'María', 'Pedro', 'Sofía']
calificaciones = [85, 90, 78, 95, 88]
plt.figure(figsize=(8, 5))
bars = plt.bar(estudiantes, calificaciones, color='skyblue', edgecolor='black')
plt.title('Calificaciones de estudiantes', fontsize=14, fontweight='bold')
plt.xlabel('Estudiante', fontsize=12)
plt.ylabel('Calificación', fontsize=12)
plt.ylim(0, 100)

for bar in bars:
    if bar.get_height() == max(calificaciones):
        bar.set_color('orange')

plt.show()
```

```python
# EJERCICIO NUMERO 3
import matplotlib.pyplot as plt

# 1. Lista vacía
pares = []

# 2. Ciclo para obtener los pares
for i in range(2, 21):
    if i % 2 == 0:
        pares.append(i)

# 3. PRIMERO: Mostrar los resultados en texto
print("Los números pares son:", pares)

# 4. LUEGO: Crear el gráfico de barras
plt.bar(pares, pares)

# 5. Mostrar el gráfico
plt.show()
```

```python
# EJERCICIO NUMERO 4
import matplotlib.pyplot as plt

# 1. Datos
dias = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes']
temps = [18, 22, 31, 27, 16]

colores = []

# 2. Clasificar y asignar colores
for t in temps:
    if t >= 30:
        print(f"{t}°C: Caluroso")
        colores.append('red')
    elif t >= 20:
        print(f"{t}°C: Templado")
        colores.append('orange')
    else:
        print(f"{t}°C: Frío")
        colores.append('blue')

# 3. Gráfico de barras
plt.bar(dias, temps, color=colores)

# 4. Título
plt.title('Temperaturas de la semana')

# 5. Mostrar
plt.show()
```

```python
# EJERCICIO NUMERO 5
import matplotlib.pyplot as plt

notas = [85, 60, 90, 45, 75, 68, 100]
aprobados = 0
reprobados = 0

for nota in notas:
    if nota >= 70:
        aprobados += 1
    else:
        reprobados += 1

plt.bar(['Aprobados', 'Reprobados'], [aprobados, reprobados])
plt.title('Quedados o +1000 de Aura')
plt.ylabel('Cantidad de Estudiantes')

# 5. Mostrar
plt.show()
```

```python
# EJERCICIO NUMERO 6
import matplotlib.pyplot as plt

# 1. Pedir el número al usuario
num = int(input("Ingresa un número: "))

# 2. Crear listas vacías
multiplicadores = []
resultados = []

# 3. Ciclo for del 1 al 10
for i in range(1, 11):
    multiplicadores.append(i)
    resultados.append(num * i)

# 4. Graficar (usando líneas y puntos como en tu imagen)
plt.plot(multiplicadores, resultados, marker='o')

# 5. Títulos
plt.title('Tabla de multiplicar')
plt.xlabel('Multiplicador')
plt.ylabel('Resultado')

# 6. Mostrar
plt.show()
```

```python
# EJERCICIO NUMERO 7
import matplotlib.pyplot as plt

# Datos (Nombres y notas de ejemplo)
nombres = ['Ana', 'Luis', 'María', 'Pedro', 'Sofía']
notas = [95, 75, 60, 85, 45]

# Ciclo para clasificar e imprimir
for i in range(len(nombres)):
    if notas[i] >= 90:
        print(f"{nombres[i]}: Excelente")
    elif notas[i] >= 70:
        print(f"{nombres[i]}: Aprobado")
    else:
        print(f"{nombres[i]}: Reprobado")

# Graficar las notas
plt.bar(nombres, notas)
plt.title('Calificaciones de estudiantes')
plt.show()
```

```python
# EJERCICIO NUMERO 8
import matplotlib.pyplot as plt

# Datos (7 días)
dias = ['Lun', 'Mar', 'Mié', 'Jue', 'Vie', 'Sáb', 'Dom']
temperaturas = [18, 25, 32, 19, 28, 35, 15]

# Ciclo para clasificar
for temp in temperaturas:
    if temp >= 30:
        print(f"{temp}°C: Calurosa")
    elif temp >= 20:
        print(f"{temp}°C: Agradable")
    else:
        print(f"{temp}°C: Fría")

# Graficar (usando líneas)
plt.plot(dias, temperaturas, marker='o')
plt.title('Temperaturas de la semana')
plt.ylabel('Temperatura (°C)')

# Mostrar
plt.show()
```

```python
# EJERCICIO NUMERO 9
import matplotlib.pyplot as plt

# 1. Datos
dias = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes']
ventas = [120000, 45000, 85000, 150000, 30000]

# 2. Ciclo simplificado con zip()
for dia, venta in zip(dias, ventas):
    if venta >= 100000:
        print(f"{dia}: Venta alta (${venta})")
    elif venta >= 50000:
        print(f"{dia}: Venta media (${venta})")
    else:
        print(f"{dia}: Venta baja (${venta})")

# 3. Gráfico y Título en dos líneas
plt.bar(dias, ventas)
plt.title('Ventas de la semana')

# 4. Mostrar
plt.show()
```

```python
# EJERCICIO NUMERO 10
import matplotlib.pyplot as plt

# 1. Datos (Lista de números enteros de ejemplo)
numeros = [10, -5, 0, 3, -2, 0, 8, -1]

# 2. Inicializar contadores
positivos = 0
negativos = 0
ceros = 0

# 3. Ciclo para clasificar y contar
for num in numeros:
    if num > 0:
        positivos += 1
    elif num < 0:
        negativos += 1
    else:
        ceros += 1

# 4. Preparar datos para el gráfico
categorias = ['Positivos', 'Negativos', 'Ceros']
cantidades = [positivos, negativos, ceros]

# 5. Crear gráfico de barras
plt.bar(categorias, cantidades)

# 6. Títulos (como en la imagen)
plt.title('Clasificación de números')
plt.xlabel('Categorías')
plt.ylabel('Cantidad')

# 7. Mostrar
plt.show()
```
